# A tour of function properties

This notebook is a guided introduction to `funcprops`. Every property example below is drawn from the same executable example corpus used by the specification tests, so the tutorial and reference documentation cannot silently drift apart.

The package answers symbolic global questions over explicit domains and assumptions. `UNKNOWN` means that no available proof route certified a conclusion; it does not mean the property is false.


```python
import sympy as sp

import funcprops as fp

x, y = sp.symbols("x y", real=True)
```

## How to read results

Simple queries return enums, SymPy sets/conditions, or structured results. Parameter-dependent questions may return guarded cases. Detailed modes expose evidence and certificates. The Markdown property reference is authoritative for definitions, return contracts, supported proof strategies, and limitations.


## Domains and values


### `boundedness`: A globally bounded function

<!-- example-id: boundedness-sine -->

**Application.** Establish stability envelopes and safe numerical search regions.

Expected result: ``Boundedness.BOUNDED``. See [`boundedness.md`](../properties/boundedness.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.boundedness(sp.sin(x), x)
assert result is fp.Boundedness.BOUNDED
result
```

### `domain`: Radical with an excluded pole

<!-- example-id: domain-radical-pole -->

**Application.** Validate the admissible input region before symbolic optimization or plotting.

Expected result: ``(x >= 1) & Ne(x, 2)``. See [`domain.md`](../properties/domain.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.domain(sp.sqrt(x - 1) / (x - 2), x)
assert sp.simplify(result ^ ((x >= 1) & sp.Ne(x, 2))) is sp.false
result
```

### `range`: Range on a restricted interval

<!-- example-id: range-restricted-quadratic -->

**Application.** Derive certified output bounds for downstream constraints.

Expected result: ``(y >= 0) & (y <= 9)``. See [`range.md`](../properties/range.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.range(x**2, x, y, constraints=sp.And(x >= -2, x <= 3))
assert sp.simplify(result ^ ((y >= 0) & (y <= 9))) is sp.false
result
```

### `sign`: Strict positivity

<!-- example-id: sign-positive -->

**Application.** Prove denominator positivity or inequality preconditions without sampling.

Expected result: ``Sign.POSITIVE``. See [`sign.md`](../properties/sign.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.sign(x**2 + 1, x)
assert result is fp.Sign.POSITIVE
result
```

### `zeros`: Exact real zeros

<!-- example-id: zeros-polynomial -->

**Application.** Locate exact roots used for sign partitions and critical-event analysis.

Expected result: ``{-2, 1}``. See [`zeros.md`](../properties/zeros.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.zeros((x - 1) * (x + 2), x)
assert result == sp.FiniteSet(-2, 1)
result
```

## Shape and variation


### `complete_monotonicity`: Mittag-Leffler complete monotonicity

<!-- example-id: complete-monotonicity-mittag-leffler -->

**Application.** Certify Laplace-transform and survival-kernel structure in fractional probability models.

Expected result: ``TruthValue.TRUE`` for ``0 < alpha <= 1``. See [`complete_monotonicity.md`](../properties/complete_monotonicity.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
alpha = sp.symbols("alpha", positive=True)
t = sp.symbols("t", nonnegative=True)
result = fp.complete_monotonicity(
    fp.MittagLefflerE(alpha, -t), t, domain=t >= 0, assumptions=alpha <= 1
)
assert result is fp.TruthValue.TRUE
result
```

## Metric regularity, variation and integrability


### `holder_regularity`: Square-root cusp exponent

<!-- example-id: holder-square-root -->

**Application.** Describe a continuous cusp where a Lipschitz bound is unavailable.

Expected result: exponent 1/2 with constant 1. See [`holder_regularity.md`](../properties/holder_regularity.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.holder_regularity(sp.sqrt(x), x, domain=sp.Interval(0, 1), return_result=True)
assert result.exponent == sp.Rational(1, 2) and result.constant == 1
result
```

### `integrability`: Separate integrability notions

<!-- example-id: integrability-compact-polynomial -->

**Application.** Check integral hypotheses without conflating local, absolute and improper convergence.

Expected result: Riemann and absolute integrability with integral 2/3. See [`integrability.md`](../properties/integrability.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.integrability(x**2, x, domain=sp.Interval(-1, 1))
assert result.riemann is fp.TruthValue.TRUE and result.absolutely_integrable is fp.TruthValue.TRUE and result.integral == sp.Rational(2, 3)
result
```

### `lipschitz`: Exact finite-domain Lipschitz constant

<!-- example-id: lipschitz-finite-optimum -->

**Application.** Bound distances in a finite sample without assuming its ambient interval is the domain.

Expected result: the optimal constant 3 on {0, 1, 2}. See [`lipschitz.md`](../properties/lipschitz.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.lipschitz(x**2, x, domain=sp.FiniteSet(0, 1, 2), return_result=True)
assert result.constant == 3 and result.value is fp.TruthValue.TRUE and result.optimal is fp.TruthValue.TRUE and result.verify()
result
```

### `uniformly_continuous`: Compact-domain uniform continuity

<!-- example-id: uniform-compact -->

**Application.** Choose a globally valid error tolerance on an explicitly bounded input region.

Expected result: TRUE on [-1, 1], while the same polynomial is not uniformly continuous on R. See [`uniformly_continuous.md`](../properties/uniformly_continuous.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.uniformly_continuous(x**2, x, domain=sp.Interval(-1, 1))
assert result is fp.TruthValue.TRUE and fp.uniformly_continuous(x**2, x) is fp.TruthValue.FALSE
result
```

### `variation`: Both jumps of a singleton spike

<!-- example-id: variation-singleton-spike -->

**Application.** Retain isolated function values when assembling Piecewise variation.

Expected result: total variation 20, counting both incident jumps. See [`variation.md`](../properties/variation.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
spike = sp.Piecewise((10, sp.Eq(x, 0)), (0, True))
result = fp.variation(spike, x, domain=sp.Interval(-1, 1), return_result=True)
assert result.total_variation == 20 and result.bounded_variation is fp.TruthValue.TRUE and result.verify()
result
```

## Statistical special functions


These examples exercise the theorem-backed statistical function layer. See [Statistical special functions](../statistical-special-functions.md) for supported families, parameter guards, and provenance.


### Regularized gamma CDF geometry

<!-- example-id: regularized-gamma-cdf -->

**Application.** Reason symbolically about Gamma-family CDFs without numerical sampling.

Expected result: strict increase on ``x > 0`` for ``a > 0``.


```python
a = sp.symbols("a", positive=True)
result = fp.monotonicity(fp.RegularizedGammaP(a, x), x, domain=x > 0)
assert result is fp.Monotonicity.STRICTLY_INCREASING
result
```

### Owen T derivative

<!-- example-id: owen-t-derivative -->

**Application.** Certify monotonicity used in skew-normal and Gaussian probability calculations.

Expected result: the exact positive derivative with respect to the second argument.


```python
h, a = sp.symbols("h a", real=True)
result = sp.diff(fp.OwenT(h, a), a)
assert result == sp.exp(-h**2 * (1 + a**2) / 2) / (2 * sp.pi * (1 + a**2))
result
```

### Generalized Marcum Q threshold monotonicity

<!-- example-id: marcum-q-threshold -->

**Application.** Analyze noncentral-tail probabilities and detection thresholds structurally.

Expected result: strict decrease in the positive threshold.


```python
m, a, b = sp.symbols("m a b", positive=True)
result = fp.monotonicity(sp.marcumq(m, a, b), b, domain=b > 0)
assert result is fp.Monotonicity.STRICTLY_DECREASING
result
```

### Mittag-Leffler complete monotonicity

<!-- example-id: mittag-leffler-complete-monotonicity -->

**Application.** Certify Laplace-transform geometry in fractional stochastic and reliability models.

Expected result: complete monotonicity for ``0 < alpha <= 1``.


```python
alpha = sp.symbols("alpha", positive=True)
t = sp.symbols("t", nonnegative=True)
result = fp.complete_monotonicity(
    fp.MittagLefflerE(alpha, -t), t, domain=t >= 0, assumptions=alpha <= 1
)
assert result is fp.TruthValue.TRUE
result
```

### Modified-Bessel ratio monotonicity

<!-- example-id: bessel-ratio-geometry -->

**Application.** Support directional-statistics and noncentral-distribution parameter geometry.

Expected result: strict increase for nonnegative ``nu`` on ``t > 0``.


```python
nu = sp.symbols("nu", nonnegative=True)
t = sp.symbols("t", positive=True)
ratio = sp.besseli(nu + 1, t) / sp.besseli(nu, t)
result = fp.monotonicity(ratio, t, domain=t > 0)
assert result is fp.Monotonicity.STRICTLY_INCREASING
result
```
