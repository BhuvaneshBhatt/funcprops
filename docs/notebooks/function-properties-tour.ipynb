# A tour of function properties

This notebook is a guided introduction to `funcprops`. Every property example below is drawn from the same executable example corpus used by the specification tests, so the tutorial and reference documentation are checked for consistency.

The package answers symbolic global questions over explicit domains and assumptions. `UNKNOWN` means that no available proof route certified a conclusion; it does not mean the property is false.


```python
import sympy as sp

import funcprops as fp

x, y = sp.symbols("x y", real=True)
```

## How to read results

Simple queries return enums, SymPy sets/conditions, or structured results. Parameter-dependent questions may return guarded cases. Detailed modes expose evidence and certificates. The Markdown property reference is authoritative for definitions, return contracts, supported proof strategies, and limitations.


## Domains and values


### `boundedness` — A globally bounded function

<!-- example-id: boundedness-sine -->

**Application.** Establish stability envelopes and safe numerical search regions.

Expected result: ``Boundedness.BOUNDED``. See [`boundedness.md`](../properties/boundedness.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.boundedness(sp.sin(x), x)
result
```

### `domain` — Radical with an excluded pole

<!-- example-id: domain-radical-pole -->

**Application.** Validate the admissible input region before symbolic optimization or plotting.

Expected result: ``(x >= 1) & Ne(x, 2)``. See [`domain.md`](../properties/domain.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.domain(sp.sqrt(x - 1) / (x - 2), x)
result
```

### `range` — Range on a restricted interval

<!-- example-id: range-restricted-quadratic -->

**Application.** Derive certified output bounds for downstream constraints.

Expected result: ``(y >= 0) & (y <= 9)``. See [`range.md`](../properties/range.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.range(x**2, x, y, constraints=sp.And(x >= -2, x <= 3))
result
```

### `sign` — Strict positivity

<!-- example-id: sign-positive -->

Expected result: ``Sign.POSITIVE``. See [`sign.md`](../properties/sign.md) for the full mathematical definition, options, proof strategies, and limitations.

### `zeros` — Exact real zeros

<!-- example-id: zeros-polynomial -->

## Shape and variation


### `complete_monotonicity` — Mittag-Leffler complete monotonicity

<!-- example-id: complete-monotonicity-mittag-leffler -->

**Application.** Certify Laplace-transform and survival-kernel structure in fractional probability models.

Expected result: ``TruthValue.TRUE`` for ``0 < alpha <= 1``. See [`complete_monotonicity.md`](../properties/complete_monotonicity.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
alpha = sp.symbols('alpha', positive=True); t = sp.symbols('t', nonnegative=True); result = fp.complete_monotonicity(fp.MittagLefflerE(alpha, -t), t, domain=t >= 0, assumptions=alpha <= 1)
result
```
