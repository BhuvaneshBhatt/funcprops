# A tour of function properties

This notebook is a guided introduction to `funcprops`. Every property example below is drawn from the same executable example corpus used by the specification tests, so the tutorial and reference documentation cannot silently drift apart.

The package answers symbolic global questions over explicit domains and assumptions. `UNKNOWN` means that no available proof route certified a conclusion; it does not mean the property is false.


```python
import sympy as sp

import funcprops as fp

x, y = sp.symbols("x y", real=True)
```

## How to read results

Simple queries return enums, SymPy sets/conditions, or structured results. Parameter-dependent questions may return guarded cases. Detailed modes expose evidence and certificates. The Markdown property reference is authoritative for definitions, return contracts, supported proof strategies, and limitations.


## Domains and values


### `function_boundedness` — A globally bounded function

<!-- example-id: boundedness-sine -->

**Application.** Establish stability envelopes and safe numerical search regions.

Expected result: ``Boundedness.BOUNDED``. See [`boundedness.md`](../properties/boundedness.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_boundedness(sp.sin(x), x)
result
```

### `function_domain` — Radical with an excluded pole

<!-- example-id: domain-radical-pole -->

**Application.** Validate the admissible input region before symbolic optimization or plotting.

Expected result: ``(x >= 1) & Ne(x, 2)``. See [`domain.md`](../properties/domain.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_domain(sp.sqrt(x - 1) / (x - 2), x)
result
```

### `function_range` — Range on a restricted interval

<!-- example-id: range-restricted-quadratic -->

**Application.** Derive certified output bounds for downstream constraints.

Expected result: ``(y >= 0) & (y <= 9)``. See [`range.md`](../properties/range.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_range(x**2, x, y, constraints=sp.And(x >= -2, x <= 3))
result
```

### `function_sign` — Strict positivity

<!-- example-id: sign-positive -->

**Application.** Prove denominator positivity or inequality preconditions without sampling.

Expected result: ``Sign.POSITIVE``. See [`sign.md`](../properties/sign.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_sign(x**2 + 1, x)
result
```

### `function_zeros` — Exact real zeros

<!-- example-id: zeros-polynomial -->

**Application.** Locate exact roots used for sign partitions and critical-event analysis.

Expected result: ``{-2, 1}``. See [`zeros.md`](../properties/zeros.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_zeros((x - 1) * (x + 2), x)
result
```

## Complex analytic structure


### `function_branchcuts` — Principal logarithm cut

<!-- example-id: branchcuts-log -->

**Application.** Avoid crossing branch cuts in symbolic transformations and complex numerical paths.

Expected result: ``(x < 0,)``. See [`branchcuts.md`](../properties/branchcuts.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_branchcuts(sp.log(x), x)
result
```

### `function_branchpoints` — Logarithmic branch point

<!-- example-id: branchpoints-log -->

**Application.** Identify non-isolated analytic structure relevant to continuation and contour choices.

Expected result: a branch point at ``x = 0``. See [`branchpoints.md`](../properties/branchpoints.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
result = fp.function_branchpoints(sp.log(x), x)
result
```

### `function_complex_zeros` — Certified complex zero in a rectangle

<!-- example-id: complex-zeros-exp -->

**Application.** Certify contour root counts before complex continuation, inversion, or spectral calculations.

Expected result: one certified zero in the rectangle. See [`complex_zeros.md`](../properties/complex_zeros.md) for the full mathematical definition, options, proof strategies, and limitations.


```python
z = sp.symbols("z")
result = fp.function_complex_zeros(
    sp.exp(z) - 1, z, rectangle=(-1 - sp.I, 1 + sp.I), return_result=True
)
result
```

### `function_discontinuities` — Jump discontinuity
